Analyze the text reviews to identify the most common positive and negative keywords.

```python
rating_counts = df["Rating text"].value_counts()
```

```python
# Visualize the distribution of rating text categories
plt.figure(figsize=(10,5))
sns.barplot(x=rating_counts.index,y=rating_counts.values)
plt.xticks(rotation=45)
plt.xlabel("Rating Category")
plt.ylabel("Count")
plt.title("Distribution of Rating Text Categories")
plt.show()
```

```python
# Diaplay most common rating 
rating_counts
```

```python
from collections import Counter
```

```python
#Define

positive_rating  = ["Excellent", "Very Good" , "Good"]
negative_rating  = ["Average","Poor"]
```

```python
#count 
positive_words = " ".join(df[df["Rating text"].isin(positive_rating)]["Rating text"]).lower().split()
negative_words = " ".join(df[df["Rating text"].isin(negative_rating)]["Rating text"]).lower().split()
```

```python
#Get most Common words

positive_word_counts = Counter(positive_words).most_common(5)
negative_word_counts = Counter(negative_words).most_common(5)

positive_word_counts,negative_word_counts

```

Calculate the average length of reviews and explore if there is a relationship between review length and rating.

```python
df["Rating the length"] = df["Rating text"].apply(lambda x: len(x.split()))
```

```python
average_length = df["Rating the length"].mean()
```

Display relation between ratingtext length and aggregate rating

```python
plt.figure(figsize=(10, 5))
sns.boxplot(
    x="Rating text", 
    y="Rating the length", 
    data=df, 
    order=rating_counts.index  # rating_counts should also match lowercase column
)
plt.xticks(rotation=45)
plt.xlabel("Rating Category")
plt.ylabel("Rating Text Length")
plt.title("Relationship Between Rating Text Length and Rating Category")
plt.show()

# Calculate average length
average_length = df["Rating the length"].mean()
print("Average length of rating text:", average_length)
```

Identify the restaurants with the highest and lowest number of votes.

```python
# Get the restaurant with the highest votes
highest_votes = df.loc[df["Votes"].idxmax()]
# Get the restaurant with the lowest votes
lowest_votes = df.loc[df["Votes"].idxmin()]
```

```python
# Display only Restaurant Name, Votes, and Aggregate rating
highest_votes[["Restaurant Name", "Votes", "Aggregate rating"]],lowest_votes[["Restaurant Name", "Votes", "Aggregate rating"]]

```

Analyze if there is a correlation between the number of votes and the rating of a restaurant.

```python
# Compute correlation between votes and aggregate rating
correlation = df["Votes"].corr(df["Aggregate rating"])
```

```python
#Visualize the relationship using a scatter plot

plt.figure(figsize=(10, 5))
sns.scatterplot(x=df["Votes"], y=df["Aggregate rating"], alpha=0.5)
plt.xlabel("Number of Votes")
plt.ylabel("Aggregate Rating")
plt.title("Correlation Between Votes and Rating")
plt.show()
```
